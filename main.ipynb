# Student Grading Dataset – Clustering Analysis

This notebook demonstrates data preprocessing and K-Means clustering on the Students Grading Dataset from Kaggle.

## 1. Import Libraries

```
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2. Generate Sample Data (Simulating Student Dataset)

```
np.random.seed(42)
data = pd.DataFrame({
    'hours_studied': np.random.uniform(0, 10, 100),
    'attendance': np.random.uniform(0.5, 1.0, 100),
    'participation': np.random.uniform(0, 1, 100),
    'previous_grades': np.random.uniform(60, 100, 100),
    'final_grade': np.random.uniform(60, 100, 100)
})
```

## 3. Data Preprocessing

```
# Remove duplicates if any
data = data.drop_duplicates()

# Normalize features
scaler = MinMaxScaler()
data_scaled = scaler.fit_transform(data)
```

## 4. Dimensionality Reduction with PCA

```
pca = PCA(n_components=2)
data_pca = pca.fit_transform(data_scaled)
```

## 5. K-Means Clustering

```
kmeans = KMeans(n_clusters=3, random_state=42)
clusters = kmeans.fit_predict(data_pca)

data['PCA1'] = data_pca[:, 0]
data['PCA2'] = data_pca[:, 1]
data['Cluster'] = clusters
```

## 6. Plotting Clusters

```
plt.figure(figsize=(8,6))
sns.scatterplot(data=data, x='PCA1', y='PCA2', hue='Cluster', palette='viridis')
plt.title('K-Means Clustering on PCA-Reduced Student Data')
plt.show()
```

## 7. Save Outputs

```
data.to_csv('student_grading_clusters.csv', index=False)
plt.savefig('kmeans_pca_plot.png')
```
